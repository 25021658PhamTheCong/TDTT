```python
#W8A3
def tong(a):
    k=0
    while(a>0):
        k+=(a%10)**2
        a=a//10
    return k
try:
    a=int(input())
    b=set()
    while(a>1):
        if a in b:
            print("NO")
            break
        b.add(a)
        a=tong(a)
    if a==1:
        print("YES")
except:
    print("loi ko mong muon")
```
