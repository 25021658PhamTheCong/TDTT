```python
#W7A1
def timkiem(lst, left, right, n):
    if left>right:
        return -1
    mid=(left + right)//2
    if lst[mid]==n:
        return mid
    elif lst[mid]>n:
        return timkiem(lst, left, mid-1, n)
    else:
        return timkiem(lst, mid+1, right, n)
lst=list(map(int,input().split()))
n=int(input())
result=timkiem(lst, 0, len(lst)-1, n)
print(result)
```

```python
#W7A3
def bubblesort(lst):
    for i in range(len(lst)):
        for j in range(0, len(lst) - i - 1):
          if lst[j] > lst[j+1]:
             lst[j], lst[j+1] = lst[j+1], lst[j]
    return lst
lst=list(map(int,input().split()))
print(bubblesort(lst))
```
