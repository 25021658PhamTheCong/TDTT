```python
#W8A1
class Cylinder :
    def __init__(self, r, h):
        self.r = r
        self.h = h
        self.pi = 3.14
    def dientich(self):
        return 2 * self.pi * (self.r**2) + 2 * self.pi * self.r *self.h
    def thetich(self):
        return self.pi * (self.r ** 2) * self.h
data = list(map(int,input().split()))
my_cylinder = Cylinder(data[0], data[1])
print(my_cylinder.dientich())
print(my_cylinder.thetich())
```

```python
#W8A2
class Date :
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
    def nextday(self):
        daymax = 0
        monthmax = 12
        month31 = [1, 3, 5, 7, 8, 10, 12]
        month30 = [4, 6, 9, 11]
        kq = f"{self.day}/{self.month}/{self.year}"
        if self.month in month31 :
            daymax = 31
            if self.day > daymax or self.month > monthmax:
                kq = "INVALID"
            elif self.day == 31:
                self.day = 1
                self.month = self.month + 1
                if self.month == 13:
                    self.month = 1
                    self.year = self.year + 1
            else:
                self.day = self.day + 1
        elif self.month in month30:
            daymax = 30
            if self.day > daymax or self.month > monthmax:
                kq = "INVALID"
            elif self.day == 30:
                self.day = 1
                self.month = self.month + 1
            else:
                self.day = self.day + 1
        elif self.month == 2:
            if self.year % 4 == 0:
                daymax = 29
                if self.day > daymax:
                    kq = "INVALID"
                elif self.day == 29:
                    self.day = 1
                    self.month = self.month + 1
                else:
                    self.day = self.day + 1
            else:
                daymax = 28
                if self.day > daymax:
                    kq = "INVALID"
                elif self.day == 28:
                    self.day = 1
                    self.month = self.month + 1
                else:
                    self.day = self.day + 1
        if kq == "INVALID":
            return kq
        else:
            kq = f"{self.day}/{self.month}/{self.year}"
        return kq
data = list(map(int,input().split("/")))
my_date = Date(data[0], data[1], data[2])
print(my_date.nextday())
```

```python
#W8A5
class ShoppingCart:
    def __init__(self):
        self.sp = []

    def add_item(self, ten, gia):
        self.sp.append({"ten": ten, "gia": gia})
        print("Đã thêm sản phẩm.")

    def remove_item(self, ten):
        for i in self.sp:
            if i["ten"] == ten:
                self.items.remove(i)
                print("Đã xóa sản phẩm.")
                return
        print("Không tìm thấy sản phẩm.")

    def is_empty(self):
        return len(self.sp) == 0

    def total_price(self):
        k = 0
        for i in self.sp:
            k += i["gia"]
        return k

    def show_items(self):
        if self.is_empty():
            print("Giỏ hàng rỗng.")
        else:
            print("Danh sách sản phẩm:")
            for i in self.sp:
                print(f"- {i['ten']} : {i['gia']}")

    def clear_cart(self):
        self.sp.clear()
        print("Đã xóa toàn bộ giỏ hàng.")
```

```python
#W8A6
class rectangle:
    def __init__(self,w,h):
        self.w=w
        self.h=h
    def area(self):
        return self.w*self.h
    def perimeter(self):
        return 2*(self.w+self.h)
    def scale(self,k):
        self.k=k
        self.w*=self.k
        self.h*=self.k
w,h,k=map(float,input().split())
x=rectangle(w,h)
x.scale(k)
print(round(x.area(),2),round(x.perimeter(),2))
```

```python
#W8A7
from math import gcd
class fraction:
    def __init__(self,num,den):
        if (den==0):
            raise ValueError("")
        if(den<0):
            num=-num
            den=-den
        g=gcd(abs(num),den)
        self.num=num//g
        self.den=den//g
    def add(self,other):
        return fraction(self.num * other.den + other.num * self.den) 
```

```python
#W8A9
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []

    def add_score(self, subject, score):
        self.scores.append(score)

    def avg(self):
        if not self.scores:
            return 0.0
        return sum(self.scores) / len(self.scores)

    def rank(self):
        average = self.avg()
        if average >= 8:
            return "Excellent"
        elif average >= 6.5:
            return "Good"
        elif average >= 5:
            return "Average"
        else:
            return "Poor"

name = input().strip()
try:
    line2 = input().strip()
    if not line2: 
        n = 0
    else:
        n = int(line2)
except ValueError:
    n = 0
student = Student(name)
for _ in range(n):
    line = input().strip().split()
    score = float(line[-1])
    subject = " ".join(line[:-1])
    student.add_score(subject, score)
print(f"{student.name} {student.avg():.2f} {student.rank()}")
```

```python
#W8A11
import math
class diem():
    def __init__(self,a=0.0,b=0.0,c=0.0):
        self.x=a
        self.y=b
        self.z=c
    def __str__(self):
        return f" {self.x}   {self.y}   {self.z} "
def tao(a):
    vt1=diem(a[1].x-a[0].x,a[1].y-a[0].y,a[1].z-a[0].z)
    vt2=diem(a[2].x-a[0].x,a[2].y-a[0].y,a[2].z-a[0].z)
    #print(vt1,vt2)
    n1=diem(vt1.y * vt2.z - vt1.z * vt2.y , vt1.z * vt2.x - vt1.x * vt2.z , vt1.x * vt2.y - vt1.y * vt2.x)
    #print(n1)
    return n1
def nhan(vt1,vt2):
    #print(vt1,vt2)
    a=abs(vt1.x*vt2.x+vt1.y*vt2.y+vt1.z*vt2.z)
    b=(vt1.x**2+vt1.y**2+vt1.z**2)**0.5*(vt2.x**2+vt2.y**2+vt2.z**2)**0.5
    #print(a,b)
    return a/b if b!=0 else 1
m1=[]
m2=[]
for i in range(4):
    a=list(map(float,input().split()))
    c=diem(a[0],a[1],a[2])
    if 0<=i<=2:
        m1.append(c)
    if 1<=i<=3:
        m2.append(c)
#print(*m1)
#print(*m2)
n1=tao(m1)
n2=tao(m2)
g=nhan(n1,n2)
#print(g)
g=abs(g)
g=math.degrees(math.acos(g))
print(f"{g:.2f}")
```
