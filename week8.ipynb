```python
#W8A1
class Cylinder :
    def __init__(self, r, h):
        self.r = r
        self.h = h
        self.pi = 3.14
    def dientich(self):
        return 2 * self.pi * (self.r**2) + 2 * self.pi * self.r *self.h
    def thetich(self):
        return self.pi * (self.r ** 2) * self.h
data = list(map(int,input().split()))
my_cylinder = Cylinder(data[0], data[1])
print(my_cylinder.dientich())
print(my_cylinder.thetich())
```

```python
#W8A2
class Date :
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
    def nextday(self):
        daymax = 0
        monthmax = 12
        month31 = [1, 3, 5, 7, 8, 10, 12]
        month30 = [4, 6, 9, 11]
        kq = f"{self.day}/{self.month}/{self.year}"
        if self.month in month31 :
            daymax = 31
            if self.day > daymax or self.month > monthmax:
                kq = "INVALID"
            elif self.day == 31:
                self.day = 1
                self.month = self.month + 1
                if self.month == 13:
                    self.month = 1
                    self.year = self.year + 1
            else:
                self.day = self.day + 1
        elif self.month in month30:
            daymax = 30
            if self.day > daymax or self.month > monthmax:
                kq = "INVALID"
            elif self.day == 30:
                self.day = 1
                self.month = self.month + 1
            else:
                self.day = self.day + 1
        elif self.month == 2:
            if self.year % 4 == 0:
                daymax = 29
                if self.day > daymax:
                    kq = "INVALID"
                elif self.day == 29:
                    self.day = 1
                    self.month = self.month + 1
                else:
                    self.day = self.day + 1
            else:
                daymax = 28
                if self.day > daymax:
                    kq = "INVALID"
                elif self.day == 28:
                    self.day = 1
                    self.month = self.month + 1
                else:
                    self.day = self.day + 1
        if kq == "INVALID":
            return kq
        else:
            kq = f"{self.day}/{self.month}/{self.year}"
        return kq
data = list(map(int,input().split("/")))
my_date = Date(data[0], data[1], data[2])
print(my_date.nextday())
```
